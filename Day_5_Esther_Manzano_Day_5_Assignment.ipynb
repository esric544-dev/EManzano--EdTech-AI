**WAP** to check job eligibility:
- Age: 21-35 (inclusive)
- If eligible, check education: "Graduate" → Interview, "Post-Graduate" → Direct Selection
- Print appropriate message for each case

```python
age = int(input("Enter Age: "))
if 21 <= age <= 35:
    edu = input("Education (Graduate/Post-Graduate): ").strip().title()
    if edu == "Post-Graduate":
        print("Status: Direct Selection")
    elif edu == "Graduate":
        print("Status: Eligible for Interview")
    else:
        print("Status: Not eligible (Education criteria)")
else:
    print("Status: Not eligible (Age criteria)")
```


**WAP** for home loan eligibility:
- Credit score >= 700
- If valid, check income: <500k (reject), 500k-1000k (3x loan), >1000k (5x loan)
- Print loan status and amount

```python
score = int(input("Enter Credit Score: "))
if score >= 700:
    income_str = input("Enter Annual Income: ")
    # Clean the input: remove commas and replace 'k'/'K' with '000'
    income_str = income_str.replace(',', '').lower().replace('k', '000')
    income = float(income_str)
    if income < 500000:
        print("Status: Rejected (Low Income)")
    elif 500000 <= income <= 1000000:
        print(f"Status: Approved! Loan Amount: ₹{income * 3}")
    else:
        print(f"Status: Approved! Loan Amount: ₹{income * 5}")
else:
    print("Status: Rejected (Low Credit Score)")
```

**WAP** to calculate ticket price:
- Age <5: Free | 5-12: ₹100 (matinee), ₹150 (evening)
- 13-59: ₹200 (matinee), ₹300 (evening) | 60+: ₹120 (matinee), ₹180 (evening)

**WAP** for ATM simulation:
- Balance: ₹10,000 | Withdraw in multiples of 100
- Charge ₹20 fee after every 3 transactions
- Loop until user enters 0

```python
balance = 10000
tx_count = 0
while True:
    amt = int(input("\nWithdraw amount (Multiples of 100, or 0 to exit): "))
    if amt == 0: break

    if amt % 100 == 0 and amt <= balance:
        balance -= amt
        tx_count += 1
        if tx_count > 3:
            balance -= 20
            print("Notice: ₹20 fee applied for frequent transactions.")
        print(f"Withdrawal Successful! Current Balance: ₹{balance}")
    else:
        print("Error: Invalid amount or insufficient balance.")
```

**WAP** for guessing game:
- Random number 1-100, max 7 attempts
- Hints: "Very close" (<5 away), "Close" (<15 away), "Too high/low"
- Declare win/loss

```python
import random
target = random.randint(1, 100)
for i in range(1, 8):
    guess = int(input(f"Attempt {i}/7 - Guess the number (1-100): "))
    diff = abs(target - guess)

    if guess == target:
        print("🎉 Congratulations! You guessed it!"); break
    elif diff < 5: print("Hint: Very close!")
    elif diff < 15: print("Hint: Close!")
    else: print("Hint: Too High" if guess > target else "Hint: Too Low")
else:
    print(f"Out of attempts! The number was {target}.")
```

**WAP** to check prime and list all divisors:
- Use while loop to find divisors
- Print "Prime" or "Composite" with divisor list

```python
num = int(input("Enter a number: "))
divisors = []
i = 1
while i <= num:
    if num % i == 0:
        divisors.append(i)
    i += 1

if len(divisors) == 2:
    print(f"{num} is a Prime Number.")
else:
    print(f"{num} is a Composite Number. Divisors: {divisors}")
```

**WAP** to analyze student marks:
- Input n students, 5 subjects each
- Calculate total, percentage, grade (A/B/C/D/F)
- If any subject <35, mark as FAIL
- Show class average

```python
n = int(input("Number of students: "))
total_class_pct = 0

for i in range(1, n + 1):
    print(f"\nStudent {i}:")
    marks = []
    for j in range(1, 6):
        while True:
            mark_input = input(f"  Subject {j} marks: ")
            try:
                mark = float(mark_input)
                marks.append(mark)
                break # Exit loop if input is valid
            except ValueError:
                print("Invalid input. Please enter a number for marks.")

    total = sum(marks)
    pct = total / 5
    total_class_pct += pct

    if any(m < 35 for m in marks): grade = "FAIL"
    elif pct >= 90: grade = "A"
    elif pct >= 75: grade = "B"
    elif pct >= 60: grade = "C"
    elif pct >= 35: grade = "D"
    else: grade = "F"

    print(f"  Total: {total}, Percentage: {pct}%, Grade: {grade}")

print(f"\nAverage Class Percentage: {total_class_pct / n:.2f}%")
```

**WAP** for shopping bill:
- Input items (name, price, quantity)
- Discount: ≥₹5000 (20%), ≥₹3000 (15%), ≥₹1000 (10%)
- Add 18% GST, print itemized bill

```python
items_count = int(input("How many different items? "))
subtotal = 0
print("-" * 20)

for _ in range(items_count):
    name = input("Item Name: ")
    price = float(input("Price: "))
    qty = int(input("Quantity: "))
    subtotal += (price * qty)

# Discount logic
if subtotal >= 5000: disc = 0.20
elif subtotal >= 3000: disc = 0.15
elif subtotal >= 1000: disc = 0.10
else: disc = 0

after_discount = subtotal * (1 - disc)
gst = after_discount * 0.18
final_total = after_discount + gst

print("\n--- ITEMIZED BILL ---")
print(f"Subtotal:      ₹{subtotal:.2f}")
print(f"Discount:     -₹{subtotal * disc:.2f}")
print(f"GST (18%):    +₹{gst:.2f}")
print(f"Final Amount:  ₹{final_total:.2f}")
```
