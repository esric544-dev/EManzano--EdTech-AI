```python
import json
import boto3

# Initialize the Bedrock client
bedrock = boto3.client(service_name='bedrock-runtime')

def lambda_handler(event, context):
    # 1. Parse the user message from the API Gateway request
    body = json.loads(event['body'])
    user_prompt = body.get('prompt', 'Hello!')

    # 2. Format the request for Amazon Bedrock (Example for Titan or Claude)
    # This shows you understand how to structure JSON for AI models
    payload = {
        "inputText": user_prompt,
        "textGenerationConfig": {
            "maxTokenCount": 512,
            "temperature": 0.7
        }
    }

    try:
        # 3. Call Amazon Bedrock
        response = bedrock.invoke_model(
            modelId='amazon.titan-text-express-v1',
            body=json.dumps(payload)
        )

        # 4. Return the AI response to the user via API Gateway
        response_body = json.loads(response['body'].read())
        return {
            'statusCode': 200,
            'body': json.dumps({'answer': response_body['results'][0]['outputText']})
        }

    except Exception as e:
        return {
            'statusCode': 500,
            'body': json.dumps({'error': str(e)})
        }
```
